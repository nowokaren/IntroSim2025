```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import glob

# Parámetros
L = 2.0
dt = 0.0001

# Listar los archivos de estado
estados = sorted(glob.glob("estados/estado_*.dat"))
n_steps = len(estados)
if n_steps == 0:
    raise RuntimeError("No se encontraron archivos en la carpeta 'estados/'")

# Leer el primer archivo para saber cuántas partículas hay
sample = np.loadtxt(estados[0])
N_part = sample.shape[0]

# Arrays para posiciones
rx_all = np.zeros((n_steps, N_part))
ry_all = np.zeros((n_steps, N_part))
rz_all = np.zeros((n_steps, N_part))

# --- Leer los archivos y aplicar condiciones periódicas ---
for i, fname in enumerate(estados):
    data = np.loadtxt(f"estados/estado_{i+1}.dat")
    #if i ==0:
    #    _, rx, ry, rz, fx, fy, fz, _ = data.T
    #else:
    #    _, _, _, _, fx, fy, fz, _ = data.T

    # Aplicar un paso de evolución + condiciones periódicas
    #rx = (rx + fx * dt**2 / 2) % L
    #ry = (ry + fy * dt**2 / 2) % L
    #rz = (rz + fz * dt**2) % L
    _, rx, ry, rz, fx, fy, fz, _ = data.T

    rx_all[i] = rx
    ry_all[i] = ry
    rz_all[i] = rz

print(f"Listo → {n_steps} pasos, {N_part} partículas")

# --- Graficar las trayectorias ---
fig = plt.figure(figsize=(9,8))
ax = fig.add_subplot(111, projection='3d')

# Cada partícula tiene su trayectoria
colors = plt.cm.jet(np.linspace(0, 1, N_part))
for j, c in enumerate(colors):
    ax.plot(rx_all[:, j], ry_all[:, j], rz_all[:, j],
            ".-", color=c, alpha = 1/(j+1))

# Dibujar caja del cubo periódico
corners = np.array([[0,0,0],[L,0,0],[L,L,0],[0,L,0],
                    [0,0,L],[L,0,L],[L,L,L],[0,L,L]])
edges = [(0,1),(1,2),(2,3),(3,0),(4,5),(5,6),(6,7),(7,4),
         (0,4),(1,5),(2,6),(3,7)]
for a,b in edges:
    ax.plot(*zip(corners[a], corners[b]), c='k', lw=1)


ax.set_title("Trayectorias de las partículas (con contorno periódico)")
plt.tight_layout()
plt.show()
```
